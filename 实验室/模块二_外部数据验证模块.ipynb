```python
#三模型对比预测
#集成模型、CV最优单一模型、权重最大单一模型 预测外部8个工况
ext_data = load_external()
print('外部验证工况数:', {k: len(v) for k, v in ext_data.items()})


train_arr = np.vstack([
    load_condition_means(t)[FEATS].values.astype(float)
    for t in SHEETS.values()
]).round(6)

ext_arr = np.vstack([
    ext_data[t][FEATS].values.astype(float)
    for t in SHEETS.values()
]).round(6)

overlap = len(
    {tuple(r) for r in train_arr.tolist()}
    & {tuple(r) for r in ext_arr.tolist()}
)
print('外部工况与训练集的因子组合重叠数:',overlap)



all_results, result_dfs, all_comparison = [], {}, {}
for tag in SHEETS.values():
    res_df, summary_row, comparison = predict_external_with_comparison(
        tag, ENSEMBLES[tag], TUNED[tag],
        BEST_SINGLE[tag], MAXW_SINGLE[tag], ext_data[tag],
        verbose=True)
    result_dfs[tag] = res_df
    all_results.append(summary_row)
    all_comparison[tag] = comparison

summary = pd.DataFrame(all_results)
cols = ['tag', 'n_ext',
        'Ensemble_R2',
        'Ensemble_RMSE', 'Ensemble_MAPE',
        'BestSingle', 'BestSingle_R2', 'BestSingle_RMSE', 'BestSingle_MAPE',
        'MaxWeightSingle', 'MaxWeightSingle_R2', 'MaxWeightSingle_RMSE', 'MaxWeightSingle_MAPE',
        'Ens_vs_Best_wins', 'Ens_vs_Best_p',
        'Ens_vs_MaxW_wins', 'Ens_vs_MaxW_p']
summary = summary[[c for c in cols if c in summary.columns]]
print('\n外部验证汇总（含精确配对符号检验）:')
display(summary)
```

```python
# 可视化
def plot_external_results(results_dict, save_prefix='external'):
    """
    results_dict: {tag: res_df}
    绘制: (1) 预测vs真实散点图 (2) 残差图
    """
    tags = list(results_dict.keys())
    n = len(tags)
    fig, axes = plt.subplots(2, n, figsize=(5*n, 9), dpi=450)
    if n == 1:
        axes = axes.reshape(2, 1)

    for idx, tag in enumerate(tags):
        res = results_dict[tag]
        y_true, y_pred = res['y_true'].values, res['y_ensemble'].values

        # --- 预测 vs 真实 ---
        ax = axes[0, idx]
        ax.scatter(y_true, y_pred, c='steelblue', edgecolors='k', s=80, zorder=3)
        lim = [min(y_true.min(), y_pred.min())*0.95, max(y_true.max(), y_pred.max())*1.05]
        ax.plot(lim, lim, 'r--', lw=1.5, label='1:1')
        ax.set_xlim(lim); ax.set_ylim(lim)
        ax.set_xlabel('真实值', fontsize=11)
        ax.set_ylabel('预测值', fontsize=11)
        ax.set_title(f'{TAG_CN[tag]} ({tag})', fontsize=12, fontweight='bold')
        ax.legend(loc='upper left')
        ax.grid(True, alpha=0.3)

        r2 = r2_score(y_true, y_pred)
        ax.text(0.05, 0.95, f'R² = {r2:.3f}', transform=ax.transAxes,
                fontsize=11, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

        # --- 残差图 ---
        ax2 = axes[1, idx]
        resid = y_pred - y_true
        ax2.bar(range(1, len(resid)+1), resid, color=['green' if r<0 else 'coral' for r in resid],
                edgecolor='k')
        ax2.axhline(0, color='black', linestyle='-', linewidth=0.8)
        ax2.set_xlabel('工况编号', fontsize=11)
        ax2.set_ylabel('预测 - 真实 (残差)', fontsize=11)
        ax2.set_title(f'{TAG_CN[tag]} 残差分布', fontsize=12, fontweight='bold')
        ax2.grid(True, alpha=0.3, axis='y')

    plt.tight_layout()
    plt.show()

plot_external_results(result_dfs, save_prefix='external')
```

```python
#保存Excel
os.makedirs(OUT_DIR, exist_ok=True)


#含三项对比
with pd.ExcelWriter(COMP_XLSX) as writer:
    for tag, df in result_dfs.items():
        df.to_excel(writer, sheet_name=tag, index=False)
    summary.to_excel(writer, sheet_name='汇总指标', index=False)
    # 逐目标对比明细
    det = []
    for tag in SHEETS.values():
        c = all_comparison[tag]
        for k in ['Ensemble', 'BestSingle', 'MaxWeightSingle']:
            if k in c:
                det.append(dict(目标=tag, 模型=k, R2=round(c[k]['R2'], 4),
                                RMSE=round(c[k]['RMSE'], 3),
                                MAPE=round(c[k]['MAPE'], 3)))
    pd.DataFrame(det).to_excel(writer, sheet_name='三模型明细', index=False)
print('已保存:', COMP_XLSX)
```
